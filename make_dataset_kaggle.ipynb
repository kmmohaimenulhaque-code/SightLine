```python
%%writefile make_dataset.py
"""Build a labelled dataset through OpenRouter. One user message per call, no system prompt, provider-default sampling.
  export OPENROUTER_API_KEY=...        # never paste it anywhere
  python make_dataset.py --models claude-opus-4-6                   # first run
  python make_dataset.py --models claude-opus-4-8 claude-opus-5     # later: same prompts.csv, new random N per call
  python make_dataset.py --dry-run                                  # prints samples + cost estimate, no key needed
Resumable: every call is appended to raw_calls.jsonl and never repeated."""
import os, sys, csv, json, math, time, random, hashlib, argparse, re, threading
from concurrent.futures import ThreadPoolExecutor
import urllib.request, urllib.error
import topics as T

SLUG = {  # competition label -> OpenRouter model id (verified against openrouter.ai/api/v1/models)
 'gpt-5.5': 'openai/gpt-5.5', 'gpt-5.6-luna': 'openai/gpt-5.6-luna', 'gpt-5.6-sol': 'openai/gpt-5.6-sol', 'gpt-5.6-terra': 'openai/gpt-5.6-terra',
 'gpt-6-astra': 'openai/gpt-6-astra', 'gemini-3.1-pro-preview': 'google/gemini-3.1-pro-preview', 'gemini-3.6-flash': 'google/gemini-3.6-flash',
 'gemini-3.7-flash': 'google/gemini-3.7-flash', 'gemini-3.8-flash': 'google/gemini-3.8-flash', 'claude-fable-5': 'anthropic/claude-fable-5',
 'claude-opus-4-6': 'anthropic/claude-opus-4.6', 'claude-opus-4-8': 'anthropic/claude-opus-4.8', 'claude-opus-5': 'anthropic/claude-opus-5',
 'claude-sonnet-5': 'anthropic/claude-sonnet-5', 'qwen3.8-max': 'qwen/qwen3.8-max-0902', 'glm-5.3': 'z-ai/glm-5.3', 'glm-5.3-flash': 'z-ai/glm-5.3-flash',
 'kimi-k3': 'moonshotai/kimi-k3'}
PROVIDERS = {'openrouter': ('https://openrouter.ai/api/v1/chat/completions', 'OPENROUTER_API_KEY', None),
             'nvidia': ('https://integrate.api.nvidia.com/v1/chat/completions', 'NVIDIA_API_KEY', {'glm-5.3': 'z-ai/glm-5.3', 'glm-5.3-flash': 'z-ai/glm-5.3-flash'})}
VERBS = [('Explain', T.EXPLAIN), ('Describe', T.DESCRIBE), ('Argue', T.ARGUE), ('Compare', T.COMPARE), ('Narrate', T.NARRATE)]
RAWF = 'raw_calls.jsonl'
TAIL = 'Keep it to one self-contained paragraph without headings or lists.'

def make_prompts(path='prompts.csv'):
    if os.path.exists(path): return list(csv.DictReader(open(path, encoding='utf-8')))
    rng = random.Random(1); cons = [c for _ in range(25) for c in T.CONSTRAINTS]; rng.shuffle(cons); rows = []; k = 0
    for v, tl in VERBS:
        for t in tl:
            k += 1; c = cons[k - 1]; rows.append(dict(prompt_id=f'p{k:03d}', task_verb=v, topic=t, constraint=c, prompt_body=f'{v} {t}. {c}. {TAIL}'))
    rng.shuffle(rows)
    for i, r in enumerate(rows): r['prompt_id'] = f'p{i+1:03d}'
    with open(path, 'w', newline='', encoding='utf-8') as f:
        w = csv.DictWriter(f, fieldnames=list(rows[0])); w.writeheader(); w.writerows(rows)
    return rows

def target_n(pid, label):  # fresh, reproducible per (prompt, label); roughly the competition's length spread (median ~100, 25% ~65, 75% ~150)
    r = random.Random(int(hashlib.sha256(f'{pid}|{label}'.encode()).hexdigest()[:12], 16))
    return int(min(250, max(40, round(r.lognormvariate(math.log(100), 0.62)))))

def call(slug, text, key, url='https://openrouter.ai/api/v1/chat/completions', max_tokens=16000):
    body = json.dumps({'model': slug, 'messages': [{'role': 'user', 'content': text}], 'max_tokens': max_tokens}).encode()  # nothing else: provider defaults
    for attempt in range(4):
        req = urllib.request.Request(url, data=body, headers={'Authorization': f'Bearer {key}', 'Content-Type': 'application/json'})
        try:
            with urllib.request.urlopen(req, timeout=600) as r: return json.load(r)
        except urllib.error.HTTPError as e:
            if e.code in (408, 429, 500, 502, 503, 504) and attempt < 3: time.sleep(5 * 2 ** attempt); continue
            return {'error': f'HTTP {e.code}: {e.read()[:300].decode(errors="replace")}'}
        except Exception as e:
            if attempt < 3: time.sleep(5 * 2 ** attempt); continue
            return {'error': repr(e)}

REFUSAL = re.compile(r"^(i can'?t|i cannot|i'm sorry|i am sorry|i apologi[sz]e|sorry,|as an ai)", re.I)
def judge(text, finish):
    w = len(text.split())
    if finish not in ('stop', 'end_turn'): return 'not_finished'
    if not text: return 'empty'
    if REFUSAL.match(text): return 'refusal'
    if '\n' in text: return 'newline'
    if re.search(r'\*\*|^#|^\s*([-*•]|\d+[.)])\s', text, re.M): return 'markdown'
    if not 40 <= w <= 250: return 'length'
    return None

def served_ok(label, served):  # accept the exact slug, or a dated/versioned variant of it
    base = SLUG[label]; s = (served or '').lower()
    return s == base or s.startswith(base + '-') or s.startswith(base + ':')

def main():
    ap = argparse.ArgumentParser(); ap.add_argument('--models', nargs='+', default=['claude-opus-4-6']); ap.add_argument('--limit', type=int, default=200); ap.add_argument('--start', type=int, default=1)
    ap.add_argument('--workers', type=int, default=4); ap.add_argument('--provider', default='openrouter', choices=list(PROVIDERS)); ap.add_argument('--max-tokens', type=int, default=16000); ap.add_argument('--dry-run', action='store_true'); ap.add_argument('--out', default='dataset.csv')
    a = ap.parse_args(); url, keyenv, own = PROVIDERS[a.provider]
    if own: SLUG.clear(); SLUG.update(own)  # provider-specific model ids
    bad = [m for m in a.models if m not in SLUG]; assert not bad, f'unknown labels {bad}; choose from {sorted(SLUG)}'
    global RAWF; RAWF = f'raw_calls_{a.provider}.jsonl'
    P = make_prompts()[a.start - 1:a.limit]
    if a.dry_run:
        for r in P[:3]: print(f"{r['prompt_id']} N={target_n(r['prompt_id'], a.models[0])}: {r['prompt_body']} (about {target_n(r['prompt_id'], a.models[0])} words)\n")
        print(f"{len(P)} prompts x {len(a.models)} models = {len(P)*len(a.models)} calls. Cost is mostly output tokens; check openrouter.ai/<model> pricing (models with default thinking can use 1-4k tokens per call)."); return
    key = os.environ.get(keyenv); assert key, f'set {keyenv} first'
    done = set(); lock = threading.Lock()
    if os.path.exists(RAWF):
        for l in open(RAWF, encoding='utf-8'):
            d = json.loads(l); done.add((d['prompt_id'], d['label']))
    jobs = [(r, m) for m in a.models for r in P if (r['prompt_id'], m) not in done]; print(len(jobs), 'calls to make', flush=True)
    def run(j):
        r, label = j; n = target_n(r['prompt_id'], label); text = f"{r['prompt_body']} (about {n} words)"
        resp = call(SLUG[label], text, key, url, a.max_tokens); ch = (resp.get('choices') or [{}])[0] if 'error' not in resp else {}
        rec = dict(prompt_id=r['prompt_id'], label=label, requested=SLUG[label], n=n, sent=text, error=resp.get('error'), served=resp.get('model'),
                   provider=resp.get('provider'), finish=ch.get('finish_reason'), text=(ch.get('message') or {}).get('content') or '', usage=resp.get('usage'))
        with lock:
            open(RAWF, 'a', encoding='utf-8').write(json.dumps(rec, ensure_ascii=False) + '\n')
            print(r['prompt_id'], label, rec['finish'] or rec['error'], flush=True)
    with ThreadPoolExecutor(a.workers) as ex: list(ex.map(run, jobs))
    rows, drops = [], {}
    for l in open(RAWF, encoding='utf-8'):
        d = json.loads(l); t = (d['text'] or '').strip()
        why = 'api_error' if d['error'] else ('served_model_mismatch' if not served_ok(d['label'], d['served']) else judge(t, d['finish']))
        if why: drops[(d['label'], why)] = drops.get((d['label'], why), 0) + 1; continue
        rows.append(dict(prompt_id=d['prompt_id'], model=d['label'], text=t, served_model_id=d['served'], target_words=d['n']))
    with open(a.out, 'w', newline='', encoding='utf-8') as f:
        w = csv.DictWriter(f, fieldnames=['prompt_id', 'model', 'text', 'served_model_id', 'target_words']); w.writeheader(); w.writerows(rows)
    print(f'\nkept {len(rows)} rows -> {a.out}; dropped: {drops}')
    for m in sorted({r['model'] for r in rows}):
        x = [r for r in rows if r['model'] == m]; wc = [len(r['text'].split()) for r in x]
        print(f"{m}: {len(x)} rows | words min/mean/max {min(wc)}/{sum(wc)//len(wc)}/{max(wc)} | em-dash share {sum('—' in r['text'] for r in x)/len(x):.0%} | served {sorted({r['served_model_id'] for r in x})}")
if __name__ == '__main__': main()
```
